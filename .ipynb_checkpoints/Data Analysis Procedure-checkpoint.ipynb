# Loading dataset and importing libraries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns


df = pd.read_csv("Data Set/Wind_turbine.csv")
```

```python
df
```

# Data Cleaning

```python
df.isnull().sum()
```

```python
for col in df.select_dtypes(include="number").columns:
    if df[col].isnull().sum()!=0:
        df[col] = df[col].fillna(df[col].mean())
```

```python
for col in df.select_dtypes(include="number").columns:
    plt.figure(figsize=(10,8))
    sns.boxplot(data=df[col])
    plt.title(f'Boxplot for {col}')
    plt.show()
```

## Statistical Outliers

```python
for col in df.select_dtypes(include="number").columns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5*IQR
    upper = Q3 + 1.5*IQR
    print(f'Upper outlier for {col}: {upper}')
    print(f'Lower outlier for {col}: {lower}')
```

```python
columns = ['Wind_speed', 'Rotor_Speed', 'Wheel_hub_temperature', 'Bearing_temperature', 'Gear_oil_temperature']

def cap_outliers(df, column):
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    df[column] = df[column].clip(lower=lower_bound, upper=upper_bound)
    return df

for col in columns:
    df = cap_outliers(df, col)
    
```

## Standardizing formats

```python
# Step 1: Convert from MM/DD/YYYY to datetime
df['date'] = pd.to_datetime(df['date'], format='%m/%d/%Y', errors='coerce')

# Step 2: Reformat to DD/MM/YYYY
df['date'] = df['date'].dt.strftime('%d/%m/%Y')
```
