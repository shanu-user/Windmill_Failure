## Statistical Outliers

```python
for col in df.select_dtypes(include="number").columns:
    plt.figure(figsize=(10,8))
    sns.boxplot(data=df[col])
    plt.title(f'Boxplot for {col}')
    plt.show()
```

```python
for col in df.select_dtypes(include="number").columns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5*IQR
    upper = Q3 + 1.5*IQR
    print(f'Upper outlier for {col}: {upper}')
    print(f'Lower outlier for {col}: {lower}')
```

# Comparing normal with target feature

```python
import seaborn as sns

for col in df.select_dtypes(include="number").columns:
    plt.figure()
    sns.boxplot(x="Failure_status", y=col, data=df)
    plt.title(f"{col} vs Failure")
    plt.show()
```

```python
columns = ['Wind_speed', 'Rotor_Speed', 'Wheel_hub_temperature', 'Bearing_temperature', 'Gear_oil_temperature']

def cap_outliers(df, column):
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    df[column] = df[column].clip(lower=lower_bound, upper=upper_bound)
    return df

for col in columns:
    df = cap_outliers(df, col)
    
```

## Standardizing formats

```python
# Step 1: Convert from MM/DD/YYYY to datetime
df['date'] = pd.to_datetime(df['date'], format='%m/%d/%Y', errors='coerce')
```

# Feature Engineering

```python
df['month'] = df['date'].dt.month


# To check operational stress periods, seasonal trends, wind patterns are time-dependent
df['year'] = df['date'].dt.year
df['day'] = df['date'].dt.day
df['hour'] = df['date'].dt.hour
df['day_of_week'] = df['date'].dt.dayofweek  # 0=Monday
```

```python
# Shows efficiency of turbine via power-to-wind-ratio

df['Power'] / (df['Wind_speed'] + 1e-5)
```

```python
# Overheating detection and failure signals

# Nacelle-Ambient

df['Nacelle_temperature'] - df['Ambient_temperature']

# Gear-Ambient
df['Gear_oil_temperature'] - df['Ambient_temperature']
```

```python
# Rotational Efficiency

df['Power'] / (df['Rotor_Speed'] + 1e-5)
```
